# Extract WorldStrat High-Resolution Dataset (`hr_dataset.tar.gz`)
### Safe Streaming Extraction for Google Drive & Colab Storage

This notebook extracts the 41.5 GB high-resolution WorldStrat archive into structured folders.

### Extraction Modes:
1. **Selective Mode (Recommended):** Extracts only the high-res folders that match the 29–30 locations already downloaded in `worldstrat_low`. This requires **< 1 GB** of space and finishes in **~2–3 minutes**, completely avoiding Colab disk overflow (`Disk Full`).
2. **Full Mode:** Extracts all global locations (requires ~85–100 GB of free Google Drive storage).

## Step 1: Mount Google Drive & Environment Setup

```python
from google.colab import drive
drive.mount('/content/drive')

import os, tarfile, shutil
from tqdm import tqdm

BASE_DIR = '/content/drive/MyDrive/Project_Folder/Datasets'
TAR_PATH = os.path.join(BASE_DIR, 'worldstrat_high/hr_dataset.tar.gz')
LOW_DIR = os.path.join(BASE_DIR, 'worldstrat_low')
OUT_DIR = os.path.join(BASE_DIR, 'worldstrat_high/extracted')
os.makedirs(OUT_DIR, exist_ok=True)

print(f'Archive Path: {TAR_PATH}')
print(f'Output Directory: {OUT_DIR}')
print(f'Archive Exists: {os.path.exists(TAR_PATH)}')
if os.path.exists(TAR_PATH):
    size_gb = os.path.getsize(TAR_PATH) / (1024**3)
    print(f'Archive Size: {size_gb:.2f} GB')
```

## Step 2: Choose Extraction Mode & Execute

Set `MODE = 'selective'` to extract only the paired locations matching your `worldstrat_low` scenes, or `MODE = 'all'` for the entire global archive.

```python
MODE = 'selective'  # Options: 'selective' or 'all'

target_locations = set()
if MODE == 'selective' and os.path.exists(LOW_DIR):
    target_locations = set(os.listdir(LOW_DIR))
    print(f'[Selective Mode] Found {len(target_locations)} target locations in worldstrat_low.')
    print('Target samples:', list(target_locations)[:5])
else:
    print('[Full Mode] Extracting all global archive locations.')

print(f'Starting extraction into: {OUT_DIR} ...')
extracted_count = 0

with tarfile.open(TAR_PATH, 'r:gz') as tar:
    for member in tqdm(tar, desc='Extracting files'):
        top_folder = member.name.split('/')[0] if '/' in member.name else member.name

        if MODE == 'selective' and top_folder not in target_locations:
            continue

        tar.extract(member, path=OUT_DIR)
        extracted_count += 1

print(f'\nSuccessfully extracted {extracted_count} items to: {OUT_DIR}')
```

## Step 3: Verify Extracted Folders and High-Res Images

```python
extracted_folders = os.listdir(OUT_DIR)
print(f'Total High-Res Folders Extracted: {len(extracted_folders)}')
print('Sample extracted folders:', extracted_folders[:5])

# Check files inside first extracted folder
if extracted_folders:
    sample_path = os.path.join(OUT_DIR, extracted_folders[0])
    if os.path.isdir(sample_path):
        print(f'Files inside {extracted_folders[0]}:', os.listdir(sample_path))
```
